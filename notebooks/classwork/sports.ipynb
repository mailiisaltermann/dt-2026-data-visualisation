```python
# can use to work with tabular data
import pandas as pd 
df = pd.read_csv("competition.csv")
df.head() # Show the first few rows
```

```python
import seaborn as sns #use this to create more clean and formatted plots
sns.boxplot(data=df, y="age")
plt.title("Ages of the athletes")
plt.show()
```

```python
plt.Figure(figsize=(8,5))
df[["age", "Points in the competition"]].boxplot()
plt.title("Points and ages in the competition")
plt.show
```

```python
avg_age= df.groupby("Points in the competition")["age"].mean()
avg_age.plot(kind="bar")
plt.title("Ages and Points in the competition")
plt.ylabel("Age")
plt.show()
```
